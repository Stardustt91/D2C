```python
import os

import requests
```

```python
def convert_to_egp_currency(from_currency: str, amount: float) -> float:
    api_key = os.environ["EXCHANGERATE_API_KEY"]
    url = f"https://v6.exchangerate-api.com/v6/{api_key}/pair/{from_currency}/EGP"
    response = requests.get(url)
    data = response.json()
    converted_amount = amount * data['conversion_rate']
    return converted_amount
```

```python
converted_amount = convert_to_egp_currency("USD", 1)
```
